## Preparing Features for Machine Learning

```python
ml_df = df.copy()
```

```python
ml_df = ml_df.drop(columns=["Car_ID", "Year"])
```

```python
ml_df.columns
```

```python
print("Rows:", ml_df.shape[0])
print("Columns:", ml_df.shape[1])
```

```python
X = ml_df.drop(columns=["Price_Lakhs"])
y = ml_df["Price_Lakhs"]
```

```python
print("Features:", X.shape)
print("Target:", y.shape)
```

## Train-Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
print("Average training price:", y_train.mean())
print("Average testing price:", y_test.mean())
```

```python
categorical_features = X_train.select_dtypes(include=["object"]).columns
numerical_features = X_train.select_dtypes(exclude=["object"]).columns

print("Categorical features:")
print(list(categorical_features))

print("\nNumerical features:")
print(list(numerical_features))
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), categorical_features)
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Processed training data:", X_train_processed.shape)
print("Processed testing data:", X_test_processed.shape)
```

```python
print("Training data type:", type(X_train_processed))
print("Testing data type:", type(X_test_processed))
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
```

```python
linear_model = LinearRegression()

random_forest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

gradient_boosting_model = GradientBoostingRegressor(
    n_estimators=100,
    random_state=42
)
```

```python
linear_model.fit(X_train_processed, y_train)

print("Linear Regression model trained successfully.")
```

```python
random_forest_model.fit(X_train_processed, y_train)

print("Random Forest model trained successfully.")
```

```python
gradient_boosting_model.fit(X_train_processed, y_train)

print("Gradient Boosting model trained successfully.")
```

```python
linear_predictions = linear_model.predict(X_test_processed)

random_forest_predictions = random_forest_model.predict(X_test_processed)

gradient_boosting_predictions = gradient_boosting_model.predict(X_test_processed)

print("Predictions generated successfully.")
```

```python
print("Actual prices:")
print(y_test.head(10).values)

print("\nLinear Regression predictions:")
print(linear_predictions[:10])

print("\nRandom Forest predictions:")
print(random_forest_predictions[:10])

print("\nGradient Boosting predictions:")
print(gradient_boosting_predictions[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
linear_mae = mean_absolute_error(y_test, linear_predictions)
linear_rmse = np.sqrt(mean_squared_error(y_test, linear_predictions))
linear_r2 = r2_score(y_test, linear_predictions)

random_forest_mae = mean_absolute_error(y_test, random_forest_predictions)
random_forest_rmse = np.sqrt(mean_squared_error(y_test, random_forest_predictions))
random_forest_r2 = r2_score(y_test, random_forest_predictions)

gradient_boosting_mae = mean_absolute_error(y_test, gradient_boosting_predictions)
gradient_boosting_rmse = np.sqrt(mean_squared_error(y_test, gradient_boosting_predictions))
gradient_boosting_r2 = r2_score(y_test, gradient_boosting_predictions)
```

```python
print("Model Performance")
print("----------------------------")

print("\nLinear Regression")
print("MAE:", linear_mae)
print("RMSE:", linear_rmse)
print("R2 Score:", linear_r2)

print("\nRandom Forest")
print("MAE:", random_forest_mae)
print("RMSE:", random_forest_rmse)
print("R2 Score:", random_forest_r2)

print("\nGradient Boosting")
print("MAE:", gradient_boosting_mae)
print("RMSE:", gradient_boosting_rmse)
print("R2 Score:", gradient_boosting_r2)
```

```python
best_model = "Random Forest"

print("Best Model:", best_model)
print("R2 Score:", random_forest_r2)
print("MAE:", random_forest_mae)
print("RMSE:", random_forest_rmse)
```

```python
model_results = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest", "Gradient Boosting"],
    "MAE": [linear_mae, random_forest_mae, gradient_boosting_mae],
    "RMSE": [linear_rmse, random_forest_rmse, gradient_boosting_rmse],
    "R2 Score": [linear_r2, random_forest_r2, gradient_boosting_r2]
})

model_results
```

```python
import joblib

joblib.dump(random_forest_model, "random_forest_model.pkl")
joblib.dump(preprocessor, "preprocessor.pkl")

print("Model saved successfully.")
print("Preprocessor saved successfully.")
```

```python
def predict_car_price(car_data):
    car_data_processed = preprocessor.transform(car_data)
    predicted_price = random_forest_model.predict(car_data_processed)
    return predicted_price[0]

print("Prediction function created successfully.")
```
